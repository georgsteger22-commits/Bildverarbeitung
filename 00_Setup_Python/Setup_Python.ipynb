# Setup Python für Bildverarbeitung

In diesem Notebook überprüfen wir, ob die Python-Umgebung für die Lehrveranstaltung **Bildverarbeitung und Mustererkennung** korrekt eingerichtet ist.
Für die Bildverarbeitung werden verschiedene Python-Bibliotheken benötigt.
Diese werden **einmalig über das Terminal** in der virtuellen Umgebung
(`.venv`) installiert.

Die benötigten Bibliotheken sind:

- **NumPy** – numerische Berechnungen und Arbeit mit Arrays
- **OpenCV** – Bildverarbeitung und Computer Vision
- **Matplotlib** – Darstellung von Bildern und Grafiken
- **Pillow** – Öffnen und Bearbeiten von Bilddateien

Die Installation erfolgt nicht innerhalb des Notebooks, sondern im
VS-Code-Terminal.

## Installation der benötigten Pakete

Öffne in VS Code das Terminal und stelle sicher, dass die virtuelle Umgebung `.venv` aktiviert ist. Die Installation muss normalerweise nur einmal pro virtueller Umgebung durchgeführt werden.

Hier könnt ihr überprüfen, ob ihr wirklich eure .venv verwendet:

````python
import sys

print("Python executable:")
print(sys.executable)

print("\nPython version:")
print(sys.version)
````

Im Terminal sollte am Anfang der Zeile ungefähr Folgendes stehen:  

```text
(.venv) PS C:\Projekte\WS2026_Bildverarbeitung>
```

Installiere anschließend die benötigten Pakete im Terminal mit:  

```text
python -m pip install numpy matplotlib pillow opencv-python
```
